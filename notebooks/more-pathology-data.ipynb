```python
import pathlib
import numpy as np

def generate_patch(img, patch_size = (800, 800)):
    bounds = img.shape

    left_coordinate = int(np.random.uniform(0, bounds[0] - patch_size[0]))
    bottom_coordinate = int(np.random.uniform(0, bounds[1] - patch_size[1]))

    return img[
        left_coordinate:left_coordinate+800,
        bottom_coordinate:bottom_coordinate+800,
    ]
```
